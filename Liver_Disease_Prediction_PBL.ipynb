```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("indian_liver_patient.csv")
print("Dataset Shape:", df.shape)
display(df.head())
```

```python
print("Dataset Information:")
df.info()
print("Statistical Summary:")
display(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())
print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
```

```python
print("Total Duplicate Rows:")
print(df.duplicated().sum())
```

```python
duplicate_rows = df[df.duplicated()]
print("Duplicate records:")
display(duplicate_rows.head(10))
```

```python
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
print("Remaining duplicates:", df.duplicated().sum())
```

```python
print(df["Dataset"].value_counts())

sns.countplot(data=df, x="Dataset")

plt.title("Liver Disease Class Distribution")
plt.xlabel("Class")
plt.ylabel("Number of Patients")
plt.show()
```

```python
print(df["Gender"].value_counts())

sns.countplot(data=df, x="Gender")

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Age"], bins=20, kde=True)

plt.title("Age Distribution of Patients")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
numerical_columns = df.select_dtypes(
    include=np.number
).columns.drop("Dataset")

plt.figure(figsize=(14, 8))

for i, column in enumerate(numerical_columns, 1):
    plt.subplot(3, 3, i)
    sns.boxplot(y=df[column])
    plt.title(column)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))

correlation = df.select_dtypes(
    include=np.number
).corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

```python
df["Gender"] = df["Gender"].map({
    "Male": 1,
    "Female": 0
})

print("Gender encoding completed!")
print(df["Gender"].value_counts(dropna=False))
```

```python
df["Dataset"] = df["Dataset"].map({
    1: 1,
    2: 0
})

print("Target encoding completed!")
print(df["Dataset"].value_counts())
```

```python
X = df.drop("Dataset", axis=1)

y = df["Dataset"]

print("Input features shape:", X.shape)
print("Target shape:", y.shape)

print("\nInput columns:")
print(X.columns.tolist())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python
from sklearn.impute import SimpleImputer
import pandas as pd

imputer = SimpleImputer(strategy="median")

X_train = pd.DataFrame(
    imputer.fit_transform(X_train),
    columns=X.columns,
    index=X_train.index
)

X_test = pd.DataFrame(
    imputer.transform(X_test),
    columns=X.columns,
    index=X_test.index
)

print("Missing values in training data:")
print(X_train.isnull().sum().sum())

print("Missing values in testing data:")
print(X_test.isnull().sum().sum())
```

```python
from imblearn.over_sampling import RandomOverSampler

print("Before balancing:")
print(y_train.value_counts())

ros = RandomOverSampler(random_state=42)

X_train_balanced, y_train_balanced = ros.fit_resample(
    X_train,
    y_train
)

print("\nAfter balancing:")
print(y_train_balanced.value_counts())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(x=y_train_balanced)

plt.title("Class Distribution After Random Oversampling")
plt.xlabel("Class (0 = No Disease, 1 = Liver Disease)")
plt.ylabel("Number of Records")

plt.show()
```

```python
print("FINAL DATA PREPROCESSING REPORT")

print("\nTraining Data Shape:")
print(X_train_balanced.shape)

print("\nTesting Data Shape:")
print(X_test.shape)

print("\nTraining Target Distribution:")
print(y_train_balanced.value_counts())

print("\nTesting Target Distribution:")
print(y_test.value_counts())

print("\nMissing Values in Training:")
print(X_train_balanced.isnull().sum().sum())

print("\nMissing Values in Testing:")
print(X_test.isnull().sum().sum())
```

Classification Model Implementation

1.Decision Tree Classifier

```python
from sklearn.tree import DecisionTreeClassifier

# Create Decision Tree model
dt_model = DecisionTreeClassifier(random_state=42)

# Train the model
dt_model.fit(X_train_balanced, y_train_balanced)

# Make predictions
dt_predictions = dt_model.predict(X_test)

print("Decision Tree Model trained successfully!")

print("First 10 predictions:")
print(dt_predictions[:10])
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(20, 10))

plot_tree(
    dt_model,
    feature_names=X.columns,
    class_names=["No Disease", "Liver Disease"],
    filled=True,
    max_depth=3,
    fontsize=9
)

plt.title("Decision Tree Visualization")
plt.show()
```

2.Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier

# Create Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train the model
rf_model.fit(X_train_balanced, y_train_balanced)

# Make predictions
rf_predictions = rf_model.predict(X_test)

print("Random Forest Model trained successfully!")

print("First 10 predictions:")
print(rf_predictions[:10])
```

3.Naive Bayes Classifier

```python
from sklearn.naive_bayes import GaussianNB

# Create Naive Bayes model
nb_model = GaussianNB()

# Train the model
nb_model.fit(X_train_balanced, y_train_balanced)

# Make predictions
nb_predictions = nb_model.predict(X_test)

print("Naive Bayes Model trained successfully!")

print("First 10 predictions:")
print(nb_predictions[:10])
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
models = {
    "Decision Tree": dt_model,
    "Random Forest": rf_model,
    "Naive Bayes": nb_model
}

predictions = {
    "Decision Tree": dt_predictions,
    "Random Forest": rf_predictions,
    "Naive Bayes": nb_predictions
}

results = []

for name, y_pred in predictions.items():
    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1-Score": f1_score(y_test, y_pred, zero_division=0)
    })

results_df = pd.DataFrame(results)

display(results_df.round(3))
```

```python
for name, y_pred in predictions.items():
    print("\n", "=" * 50)
    print(name)
    print("=" * 50)
    print(classification_report(
        y_test,
        y_pred,
        target_names=["No Disease", "Liver Disease"],
        zero_division=0
    ))
```

```python
for name, y_pred in predictions.items():
    cm = confusion_matrix(y_test, y_pred, labels=[0, 1])

    plt.figure(figsize=(5, 4))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["No Disease", "Liver Disease"],
        yticklabels=["No Disease", "Liver Disease"]
    )

    plt.title(name + " - Confusion Matrix")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()
```

```python
print("MODEL PERFORMANCE COMPARISON")
display(results_df.round(3))
```

```python
metrics = ["Accuracy", "Precision", "Recall", "F1-Score"]

results_df.set_index("Model")[metrics].plot(
    kind="bar",
    figsize=(10, 6),
    ylim=(0, 1),
    width=0.8
)

plt.title("Performance Comparison of Machine Learning Models")
plt.ylabel("Score")
plt.xlabel("Machine Learning Models")
plt.xticks(rotation=0)
plt.legend(loc="lower right")
plt.grid(axis="y")
plt.show()
```

```python
best_model = results_df.loc[
    results_df["F1-Score"].idxmax()
]

print("BEST PERFORMING MODEL")
print("---------------------")
print("Model:", best_model["Model"])
print("Accuracy:", round(best_model["Accuracy"], 3))
print("Precision:", round(best_model["Precision"], 3))
print("Recall:", round(best_model["Recall"], 3))
print("F1-Score:", round(best_model["F1-Score"], 3))
```

```python
print("PREDICTION PROBABILITIES")
print("=" * 45)

for name, model in models.items():
    probabilities = model.predict_proba(sample_patient)[0]

    print("\n", name)
    print("No Disease Probability:",
          round(probabilities[0] * 100, 2), "%")
    print("Liver Disease Probability:",
          round(probabilities[1] * 100, 2), "%")
```
